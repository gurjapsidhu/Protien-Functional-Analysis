```python
train_df = data[data["split"] == "train"].drop_duplicates("sequence")

test_df = data[data["split"] == "test"].drop_duplicates("sequence")

common = pd.merge(
    train_df,
    test_df,
    on="sequence",
    suffixes=("_train", "_test")
)

print("Common sequences:", len(common))

print("Same GO annotations:",
      (common["go_terms_train"] == common["go_terms_test"]).sum())

print("Different GO annotations:",
      (common["go_terms_train"] != common["go_terms_test"]).sum())
```

```python
different = common[
    common["go_terms_train"] != common["go_terms_test"]
]

different[["id_train", "go_terms_train",
           "id_test", "go_terms_test"]].head()
```

```python
common["annotation_overlap"] = common.apply(
    lambda row: len(set(row["go_terms_train"]) & set(row["go_terms_test"])),
    axis=1
)

common["total_unique_terms"] = common.apply(
    lambda row: len(set(row["go_terms_train"]) | set(row["go_terms_test"])),
    axis=1
)

common["jaccard_similarity"] = (
    common["annotation_overlap"] / common["total_unique_terms"]
)

common[["id_train", "id_test", "jaccard_similarity"]].head(10)
```

```python
amino_acids = [
    'A', 'C', 'D', 'E', 'F',
    'G', 'H', 'I', 'K', 'L',
    'M', 'N', 'P', 'Q', 'R',
    'S', 'T', 'V', 'W', 'Y'
]
```

```python
sequence=data["sequence"][0]
```

```python
print(sequence)
```

```python
features = []

for aa in amino_acids:
    count = sequence.count(aa)
    frequency = count / len(sequence)
    features.append(frequency)

print(features)
```

```python
print(sum(features))
```

```python
def calculate_aac(sequence):
    features = []

    for aa in amino_acids:
        frequency = sequence.count(aa) / len(sequence)
        features.append(frequency)

    return features
```

```python
X = data["sequence"].apply(calculate_aac)

X = pd.DataFrame(X.tolist(), columns=amino_acids)

print(X.shape)
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
mlb = MultiLabelBinarizer()

Y = mlb.fit_transform(data["go_terms"])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
lg=LogisticRegression(max_iter=1000)
rf=RandomForestClassifier()
xg=XGBClassifier()
```

```python
import numpy as np
from scipy.sparse import csr_matrix
from sklearn.linear_model import SGDClassifier
from sklearn.multiclass import OneVsRestClassifier
from sklearn.preprocessing import normalize

# 1. Keep X in float32 and Y in sparse CSR format to save RAM
X_norm = normalize(X_train.astype(np.float32))
Y_sparse = csr_matrix(Y_train)

# 2. Pure ML Logistic/SGD model tuned for rare GO terms
base_ml = SGDClassifier(
    loss="log_loss",             # Outputs probability scores (predict_proba)
    penalty="l2",
    alpha=1e-5,                  # Lighter regularization so specific child GO terms appear
    class_weight="balanced",     # Boosts rare classes (Transporter, Structural, Kinase)
    max_iter=1000,
    tol=1e-3,
    n_jobs=1,                    # Keep 1 here; parallelize at OneVsRest level
    random_state=42
)

# Use n_jobs=4 (safe for i5 4-core/8-thread + 8 GB RAM without freezing)
model_v2 = OneVsRestClassifier(base_ml, n_jobs=4,verbose=10)
model_v2.fit(X_norm, Y_sparse)

# 3. Inference with 0.25 threshold so specific child GO terms show up
def predict_go_ml(seq_vector, threshold=0.25):
    vec = normalize(seq_vector.reshape(1, -1).astype(np.float32))
    probs = model_v2.predict_proba(vec)
    hits = np.where(probs >= threshold)
    return sorted([(int(i), float(probs[i])) for i in hits], key=lambda x: -x)

```
